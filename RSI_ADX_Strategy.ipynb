```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt

## Computation for RSI
def compute_rsi(close, window=14):
    delta = close.diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)

    avg_gain = gain.rolling(window).mean()
    avg_loss = loss.rolling(window).mean()

    rs = avg_gain / avg_loss
    rsi = 100 - (100 / (1 + rs))
    return rsi

## Computation for ADX
def compute_adx(high, low, close, window=14):
    plus_dm = high.diff()
    minus_dm = low.diff().abs()

    plus_dm[plus_dm < 0] = 0
    minus_dm[minus_dm < 0] = 0

    tr = pd.concat([
        high - low,
        (high - close.shift()).abs(),
        (low - close.shift()).abs()
    ], axis=1).max(axis=1)

    atr = tr.rolling(window).mean()

    plus_di = 100 * (plus_dm.rolling(window).mean() / atr)
    minus_di = 100 * (minus_dm.rolling(window).mean() / atr)

    dx = (abs(plus_di - minus_di) / (plus_di + minus_di)) * 100
    adx = dx.rolling(window).mean()

    return adx

## Feature engineering
def prepare_features(df):
    df = df.copy()

    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    df["RSI"] = compute_rsi(df["Close"])
    df["ADX"] = compute_adx(df["High"], df["Low"], df["Close"])
    df["MA"] = df["Close"].rolling(50).mean()
    df["MA_slope"] = df["MA"].diff()

    return df

## Strategy (RSI-only)
def rsi_only_strategy(df):
    signal = pd.Series(0, index=df.index)

    long_cond = (df["RSI"] > 55) & (df["MA_slope"] > 0)
    short_cond = (df["RSI"] < 45) & (df["MA_slope"] < 0)

    signal[long_cond] = 1
    signal[short_cond] = -1

    return signal

## Strategy (RSI with ADX)
def rsi_adx_strategy(df, adx_thresh=25):
    signal = pd.Series(0, index=df.index)

    long_cond = (
        (df["ADX"] > adx_thresh) &
        (df["RSI"] > 55) &
        (df["MA_slope"] > 0)
    )

    short_cond = (
        (df["ADX"] > adx_thresh) &
        (df["RSI"] < 45) &
        (df["MA_slope"] < 0)
    )

    signal[long_cond] = 1
    signal[short_cond] = -1

    return signal

## Backtestign engine
def backtest(df, signal, cost_bps=5):
    df = df.copy()
    cost = cost_bps / 10000

    df["signal"] = signal
    df["position"] = df["signal"].shift()
    df["position"].fillna(0, inplace=True)

    df["return"] = df["Close"].pct_change()
    df["strategy_return"] = df["position"] * df["return"]

    trades = df["position"].diff().abs()
    df["cost"] = trades * cost
    df["net_return"] = df["strategy_return"] - df["cost"]

    df["equity"] = (1 + df["net_return"]).cumprod()

    return df

## Performance Metrics
def performance_stats(df):
    rets = df["net_return"].dropna()

    cagr = (df["equity"].iloc[-1]) ** (252 / len(df)) - 1
    sharpe = np.sqrt(252) * rets.mean() / rets.std()
    max_dd = (df["equity"] / df["equity"].cummax() - 1).min()
    hit_rate = (rets > 0).mean()
    turnover = df["position"].diff().abs().sum() / len(df)

    return {
        "CAGR": round(cagr, 3),
        "Sharpe": round(sharpe, 3),
        "Max Drawdown": round(max_dd, 3),
        "Hit Rate": round(hit_rate, 3),
        "Turnover": round(turnover, 3)
    }

## Download and use tickers
tickers = ["TSLA", "NVDA", "IREN"]
start_date = "2018-01-01"

results = {}

for ticker in tickers:
    print(f"\nRunning backtests for {ticker}")

    data = yf.download(ticker, start=start_date, auto_adjust=True)
    data = prepare_features(data)

    # RSI-only
    rsi_signal = rsi_only_strategy(data)
    rsi_bt = backtest(data, rsi_signal)

    # RSI + ADX
    rsi_adx_signal = rsi_adx_strategy(data)
    rsi_adx_bt = backtest(data, rsi_adx_signal)

    results[ticker] = {
        "RSI-only": performance_stats(rsi_bt),
        "RSI + ADX": performance_stats(rsi_adx_bt)
    }

    # Plot equity curves
    plt.figure(figsize=(10, 4))
    plt.plot(rsi_bt["equity"], label="RSI-only")
    plt.plot(rsi_adx_bt["equity"], label="RSI + ADX")
    plt.title(f"{ticker} – Strategy Comparison")
    plt.legend()
    plt.grid(True)
    plt.show()

## Print results
for ticker, stats in results.items():
    print(f"\n{ticker} Performance:")
    for strat, metrics in stats.items():
        print(f"  {strat}: {metrics}")
```
